```python
from sklearn.model_selection import train_test_split, GridSearchCV #untuk split data into training amd testing sets #Grid Search Cross Validation on parameter turning
from sklearn.preprocessing import LabelEncoder #to transforming the categorical features into numeric ones
from sklearn.metrics import mean_squared_error, r2_score #on model evolation 
from sklearn.linear_model import LinearRegression, Ridge #on modeling
from sklearn.ensemble import RandomForestRegressor #medeling
from sklearn.tree import DecisionTreeRegressor #modeling
```

```python
df.columns
```

```python
df.head(2)
```

```python
features = ["study_hours_per_day", "attendance_percentage", "mental_health_rating", "sleep_hours", "part_time_job"]
```

```python
target = "exam_score"
```

```python
df_model = df[features + [target]].copy()
```

```python
df_model
```

***encoding technique*** 

```python
le = LabelEncoder()
```

```python
#No=0 & Yes=1
#because in alphabetical N coming first before the Y
le.fit_transform(df_model["part_time_job"])
```

```python
df_model["part_time_job"] = le.fit_transform(df_model["part_time_job"])
```

```python
df_model
```

***split the data***


```python
X = df_model[features]
```

```python
y = df_model[target]
```

```python
#testing set for testing the model
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2)
#testing_size = jumlah data yang akan di testing (testing set) dan sisanya akan di training (training set)
```

```python
len(y_test)
```

```python
len(y_train)
```

```python
models = {
    "LinearRegression":{
        "model": LinearRegression(),
        "params": {}
    },
    "DecisionTree" : {
        "model" : DecisionTreeRegressor(),
        "params" : {"max_depth": [3,5,10], "min_samples_split": [2,5]}

    },
    "RandomForestRegressor": {
        "model": RandomForestRegressor(), 
        "params": {"n_estimators": [50, 100], "max_depth": [5, 10]}
    }
}
```

```python
best_models = []
```

```python
for name, config in models.items():
    print(f"Training {name}")

    grid = GridSearchCV(config["model"], config["params"],cv=5, scoring ="neg_mean_squared_error")
    grid.fit(X_train, y_train)

    y_pred = grid.predict(X_test)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    best_models.append({
        "model":name,
        "best_params" : grid.best_params_,
        "rmse" : rmse,
        "r2" : r2
    })
```

```python
best_models
```

```python
results_df = pd.DataFrame(best_models)
```

```python
results_df
```

```python
results_df.sort_values(by="rmse")
```

```python
import joblib

best_row = results_df.sort_values(by="rmse").iloc[0]
```

```python
best_row
```

```python
best_model_name = best_row["model"]
```

```python
best_model_name
```

```python
best_model_config = models[best_model_name]
```

```python
best_model_config
```

```python
final_model = best_model_config["model"]
```

```python
final_model = best_model_config["model"].set_params(**best_row["best_params"])
```

```python
final_model.fit(X,y)
```

```python
final_model.predict(X_test)
```

```python
joblib.dump(final_model, "best_model.pkl")
```
